# HaluEval — Llama 3.2 1B Evaluation

**Model:** `meta-llama/Llama-3.2-1B-Instruct` via Ollama (`llama3.2:1b`, 1.2B params)

Llama 3.2 1B is instruction-tuned and strong at **summarization**, **short writing**, and **fact extraction** from context. It struggles with **multi-hop reasoning** — HaluEval QA (HotpotQA) is the hardest task here. Expect better scores on summarization and dialogue.

**Metrics:**
- Answer generation — similarity to ground truth vs hallucinated answer
- Hallucination detection — HaluEval Yes/No benchmark

```python
import json
import random
import re
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import requests
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, classification_report
from sklearn.metrics.pairwise import cosine_similarity
from tqdm.notebook import tqdm
```

```python
DATA_DIR = Path("../data")
OLLAMA_URL = "http://localhost:11434"
MODEL = "llama3.2:1b"
SAMPLE_SIZE = 30
TASKS = ["qa", "dialogue", "summarization", "general"]

SYSTEM_PROMPT = (
    "You are Llama 3.2 1B. You excel at concise writing, summarization, "
    "and extracting facts from provided text. Ground answers in context. Be brief."
)

TASK_OPTS = {
    "qa": {"temperature": 0.1, "num_predict": 64},
    "dialogue": {"temperature": 0.3, "num_predict": 128},
    "summarization": {"temperature": 0.2, "num_predict": 150},
    "general": {"temperature": 0.4, "num_predict": 128},
}
```

```python
def load_jsonl(path):
    rows = []
    for line in path.read_text(encoding="utf-8").splitlines():
        if line.strip():
            rows.append(json.loads(line))
    return rows

def normalize(task, row):
    if task == "qa":
        return {"question": row["question"], "context": row.get("knowledge", ""),
                "ground_truth": row["right_answer"], "hallucinated": row["hallucinated_answer"]}
    if task == "dialogue":
        return {"question": row.get("dialogue_history", ""), "context": row.get("knowledge", ""),
                "ground_truth": row["right_response"], "hallucinated": row["hallucinated_response"]}
    if task == "summarization":
        return {"question": "Summarize this document.", "context": row.get("document", ""),
                "ground_truth": row["right_summary"], "hallucinated": row["hallucinated_summary"]}
    label = row.get("hallucination_label", "")
    resp = row.get("chatgpt_response", "")
    return {"question": row["user_query"], "context": "",
            "ground_truth": resp if label == "No" else "",
            "hallucinated": resp if label == "Yes" else ""}

def load_task(task, n=SAMPLE_SIZE):
    path = DATA_DIR / f"{task}_data.json"
    return [normalize(task, r) for r in load_jsonl(path)[:n]]
```

```python
def ask_llama(prompt, task="qa"):
    opts = TASK_OPTS.get(task, TASK_OPTS["general"])
    r = requests.post(f"{OLLAMA_URL}/api/chat", json={
        "model": MODEL,
        "messages": [
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": prompt},
        ],
        "stream": False,
        "options": {"temperature": opts["temperature"], "num_predict": opts["num_predict"]},
    }, timeout=180)
    r.raise_for_status()
    return r.json()["message"]["content"].strip()

def text_sim(a, b):
    if not a or not b:
        return 0.0
    vec = TfidfVectorizer().fit_transform([a.lower(), b.lower()])
    return float(cosine_similarity(vec[0:1], vec[1:2])[0][0])

def answer_prompt(sample, task):
    ctx, q = sample["context"], sample["question"]
    if task == "summarization":
        return f"Summarize in 2-3 sentences. Only use facts from the text.\n\nDocument:\n{ctx}"
    if task == "dialogue":
        return f"Write the next dialogue response using only the knowledge.\n\nKnowledge:\n{ctx}\n\nDialogue:\n{q}\n\nResponse:"
    if task == "qa" and ctx:
        return f"Read the context. Answer with a short phrase only.\n\nContext:\n{ctx}\n\nQuestion: {q}\n\nAnswer:"
    return f"Question: {q}\n\nAnswer briefly."

def classify_error(answer, sample):
    st, sh = text_sim(answer, sample["ground_truth"]), text_sim(answer, sample["hallucinated"])
    if st > 0.5:
        return "correct"
    if sh > 0.5:
        return "fabricated_answer"
    if st > sh:
        return "minor_factual_error"
    return "unsupported_claim"
```

## 1. Answer Generation Evaluation

```python
answer_results = []

for task in TASKS:
    for sample in tqdm(load_task(task), desc=f"answer/{task}"):
        if not sample["ground_truth"]:
            continue
        answer = ask_llama(answer_prompt(sample, task), task)
        st = text_sim(answer, sample["ground_truth"])
        sh = text_sim(answer, sample["hallucinated"])
        correct = st > sh and st > 0.3
        answer_results.append({
            "task": task, "correct": int(correct),
            "sim_truth": round(st, 3), "sim_hallucinated": round(sh, 3),
            "error_type": classify_error(answer, sample),
            "question": sample["question"][:80],
            "answer": answer[:120],
        })

df_answers = pd.DataFrame(answer_results)
df_answers
```

```python
answer_metrics = df_answers.groupby("task").agg(
    accuracy=("correct", "mean"),
    avg_sim_truth=("sim_truth", "mean"),
    avg_sim_hallucinated=("sim_hallucinated", "mean"),
    n=("correct", "count"),
).round(3)

answer_metrics["error_rate"] = (1 - answer_metrics["accuracy"]).round(3)
answer_metrics["hallucination_rate"] = df_answers.groupby("task")["error_type"].apply(
    lambda x: (x == "fabricated_answer").mean()
).round(3)

print("=== Answer Generation Metrics ===")
display(answer_metrics)

print("\n=== Error Breakdown ===")
display(pd.crosstab(df_answers["task"], df_answers["error_type"]))
```

## 2. Hallucination Detection (HaluEval Benchmark)

```python
def parse_yes(text):
    return bool(re.search(r"\byes\b", text.lower()))

detection_results = []

for task in TASKS:
    for sample in tqdm(load_task(task), desc=f"detect/{task}"):
        if not sample["ground_truth"] or not sample["hallucinated"]:
            continue
        is_hall = random.choice([True, False])
        candidate = sample["hallucinated"] if is_hall else sample["ground_truth"]
        prompt = (
            f"Context:\n{sample['context']}\n\nResponse:\n{candidate}\n\n"
            "Does this response contain hallucination? Answer Yes or No only."
        )
        judgment = ask_llama(prompt, task)
        predicted = parse_yes(judgment)
        detection_results.append({
            "task": task, "is_hallucinated": int(is_hall),
            "predicted": int(predicted), "judgment": judgment[:60],
        })

df_detect = pd.DataFrame(detection_results)
df_detect
```

```python
detection_rows = []

for task in TASKS:
    sub = df_detect[df_detect["task"] == task]
    if sub.empty:
        continue
    y_true, y_pred = sub["is_hallucinated"], sub["predicted"]
    p, r, f1, _ = precision_recall_fscore_support(y_true, y_pred, average="binary", zero_division=0)
    detection_rows.append({
        "task": task,
        "accuracy": round(accuracy_score(y_true, y_pred), 3),
        "precision": round(p, 3),
        "recall": round(r, 3),
        "f1": round(f1, 3),
        "hallucination_rate": round(y_true.mean(), 3),
        "n": len(sub),
    })

df_detection_metrics = pd.DataFrame(detection_rows).set_index("task")
print("=== Hallucination Detection Metrics ===")
display(df_detection_metrics)

print("\n=== Overall Classification Report ===")
print(classification_report(df_detect["is_hallucinated"], df_detect["predicted"], target_names=["truth", "hallucinated"]))
```

## 3. Visualizations

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

answer_metrics["accuracy"].plot(kind="bar", ax=axes[0], color="#3b6ef5", edgecolor="white")
axes[0].set_title("Answer Accuracy by Task")
axes[0].set_ylabel("Accuracy")
axes[0].set_ylim(0, 1)
axes[0].tick_params(axis="x", rotation=0)

df_detection_metrics["f1"].plot(kind="bar", ax=axes[1], color="#4ade80", edgecolor="white")
axes[1].set_title("Hallucination Detection F1 by Task")
axes[1].set_ylabel("F1")
axes[1].set_ylim(0, 1)
axes[1].tick_params(axis="x", rotation=0)

plt.tight_layout()
plt.savefig("../results/metrics.png", dpi=150)
plt.show()
```

```python
out = Path("../results")
out.mkdir(exist_ok=True)
answer_metrics.to_csv(out / "answer_metrics.csv")
df_detection_metrics.to_csv(out / "detection_metrics.csv")
df_answers.to_csv(out / "answer_details.csv", index=False)
df_detect.to_csv(out / "detection_details.csv", index=False)
print(f"Saved results to {out.resolve()}")
```
